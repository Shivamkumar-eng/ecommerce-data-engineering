```python
print("CUSTOMERS")
print(customers.isnull().sum())

print("\nPRODUCTS")
print(products.isnull().sum())

print("\nORDERS")
print(orders.isnull().sum())

print("\nPAYMENTS")
print(payments.isnull().sum())
```

```python
print("Customer duplicates:", customers.duplicated().sum())
print("Product duplicates:", products.duplicated().sum())
print("Order duplicates:", orders.duplicated().sum())
print("Payment duplicates:", payments.duplicated().sum())
```

```python
print("Duplicate Customer IDs:", customers["customer_id"].duplicated().sum())

print("Duplicate Product IDs:", products["product_id"].duplicated().sum())

print("Duplicate Order IDs:", orders["order_id"].duplicated().sum())

print("Duplicate Payment IDs:", payments["payment_id"].duplicated().sum())
```

```python
# Invalid customer IDs in orders
invalid_customer_ids = orders[
    ~orders["customer_id"].isin(customers["customer_id"])
]

print("Orders with invalid customer IDs:", len(invalid_customer_ids))


# Invalid product IDs in orders
invalid_product_ids = orders[
    ~orders["product_id"].isin(products["product_id"])
]

print("Orders with invalid product IDs:", len(invalid_product_ids))


# Invalid order IDs in payments
invalid_order_ids = payments[
    ~payments["order_id"].isin(orders["order_id"])
]

print("Payments with invalid order IDs:", len(invalid_order_ids))
```

```python
# Negative or zero product prices
invalid_prices = products[products["price"] <= 0]

print("Products with invalid prices:", len(invalid_prices))


# Missing product categories
print(
    "Products with missing categories:",
    products["category"].isnull().sum()
)


# Invalid order quantities
invalid_quantities = orders[orders["quantity"] <= 0]

print(
    "Orders with invalid quantities:",
    len(invalid_quantities)
)


# Missing customer emails
print(
    "Customers with missing emails:",
    customers["email"].isnull().sum()
)


# Missing payment methods
print(
    "Payments with missing payment methods:",
    payments["payment_method"].isnull().sum()
)
```
